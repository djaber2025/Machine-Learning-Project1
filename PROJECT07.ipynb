### using the pipeline

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, TargetEncoder
from sklearn.metrics import r2_score, mean_absolute_error
from xgboost import XGBRegressor

# 1. قراءة وتنظيف أساسي
data = pd.read_csv('car_price_prediction.csv')
data.drop_duplicates(inplace=True)

# استخراج ميزة Turbo قبل مسح الكلمة
data['Is_Turbo'] = data['Engine volume'].astype(str).str.contains('Turbo').astype(int)

# تحويل النصوص لأرقام
data['Levy'] = data['Levy'].replace('-', np.nan).astype(float)
data['Mileage'] = data['Mileage'].astype(str).str.replace('km', '').astype(float)
data['Engine volume'] = data['Engine volume'].astype(str).str.replace('Turbo', '').astype(float)

# 2. تنحية الأسعار الوهمية (أقل من 1500$) + الـ Outliers الفوقانية
data = data[(data['Price'] >= 1500) & (data['Price'] <= data['Price'].quantile(0.95))]
data = data[data['Mileage'] <= data['Mileage'].quantile(0.98)]

# 3. Feature Engineering (صناعة ميزات ذكية)
data['Car_Age'] = 2021 - data['Prod. year']
data['Mileage_per_year'] = data['Mileage'] / (data['Car_Age'] + 1)

# 4. تحديد X و Y (خلينا Levy و Prod. year!)
Y = data['Price']
X = data.drop(columns=['ID', 'Price', 'Color', 'Doors'])

# 5. تقسيم البيانات
X_train, X_temp, Y_train, Y_temp = train_test_split(X, Y, test_size=0.3, random_state=42)
X_val, X_test, Y_val, Y_test = train_test_split(X_temp, Y_temp, test_size=0.5, random_state=42)

# 6. Pipeline احترافي بـ TargetEncoder (ملاحظة: XGBoost ما يحتاجش StandardScaler!)
num_cols = ['Levy', 'Prod. year', 'Engine volume', 'Mileage', 'Cylinders', 'Airbags', 'Is_Turbo', 'Car_Age', 'Mileage_per_year']
low_cat_cols = ['Wheel', 'Gear box type', 'Drive wheels', 'Leather interior', 'Fuel type', 'Category']
high_cat_cols = ['Manufacturer', 'Model']

preprocessor = ColumnTransformer([
    ('num', 'passthrough', num_cols), # XGBoost يتعامل مع الأرقام والـ NaN مباشرة!
    ('low_cat', OneHotEncoder(handle_unknown='ignore', drop='first'), low_cat_cols),
    ('high_cat', TargetEncoder(smooth='auto', random_state=42), high_cat_cols)
])

model_pipeline = Pipeline([
    ('prep', preprocessor),
    ('model', XGBRegressor(
        n_estimators=600,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=5.0,
        random_state=42,
        n_jobs=-1
    ))
])

# 7. التدريب والتقييم
model_pipeline.fit(X_train, Y_train)

train_score = r2_score(Y_train, model_pipeline.predict(X_train))
val_score = r2_score(Y_val, model_pipeline.predict(X_val))
mae = mean_absolute_error(Y_val, model_pipeline.predict(X_val))

print(f"Train R2 : {train_score * 100:.2f} %")
print(f"Val R2   : {val_score * 100:.2f} %")
print(f"MAE Error: {mae:.2f} $")
```

```python
test_preds = model_pipeline.predict(X_test)
test_score = r2_score(Y_test, test_preds)
test_mae = mean_absolute_error(Y_test, test_preds)

print("-" * 35)
print(f"Final Test R2  : {test_score * 100:.2f} %")
print(f"Final Test MAE : {test_mae:.2f} $")
```
